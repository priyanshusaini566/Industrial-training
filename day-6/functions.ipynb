```python
# basic calculator

def add(x,y):
    sum=x+y
    print("addition of ",x," and ",y,"is : ",sum)

def sub(x,y):
    minus=x-y
    print(f"subtraction will be :  ={minus}")

def mul(x,y):
    new=x*y
    print("multiplication will be : ={}".format(new))

def div(x,y):
    again=x/y
    print("division  of ",x," and ",y,"is : ",again)

x=int(input("Enter 1st number : "))
y=int(input("Enter 2nd number : "))
op=input("Enter the operation : ")

if op=='+':
    add(x,y)
if op=='-':
    sub(x,y)
if op=='*':
    mul(x,y)
if op=='/':
    div(x,y)
```
