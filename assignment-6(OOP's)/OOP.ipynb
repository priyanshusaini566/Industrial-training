```python
"""
Create a class Employee with private variable _age.

Write a setter method that accepts only age greater than 18.
If user tries to set less than 18, print “Invalid age”.
Write a getter method to print the age.
"""
class Employee:
    def __init__(self):
        self._age=None
    def setter(self,age):
        if age<18:
            print("invalid age")
        else:
            self._age=age

    def getter(self):
        print(self._age)

e1=Employee()
age=int(input("enter the age : "))
e1.setter(age)
e1.getter()
```

# question-4

```python
"""
Create a class Product with _price and _discount.

Use set_price() and set_discount() to assign values.
Write a get_final_price() method to return final price after discount. Formula: price - (price * discount / 100)
Display the final price.
"""

class Product:
    def set_price(self,amt):
        self._amt=amt
    def set_discount(self,discount):
        self._discount=discount
    def get_final_price(self):
        return self._amt-((self._amt*self._discount)/100)
    
p1=Product()
price=int(input("enter the price : "))
discount=int(input("enter the discount in % : "))
p1.set_price(price)
p1.set_discount(discount)
print("the final price will be : ",p1.get_final_price())
    
```

```python
"""
Create a class Temperature with private variable _celsius.

Write set_celsius() and get_celsius() for direct values.
Write set_fahrenheit() to accept Fahrenheit and convert to Celsius. Formula: (F - 32) * 5/9
Write get_fahrenheit() to return Fahrenheit value.
Test both methods.
"""
class Temperature:
    def set_celsius(self,celsius):
        self._celsius=celsius
    def get_celsius(self):
        return (self._celsius)
        

    def set_fahrenheit(self,f):
         self._celsius=(f-32)*(5/9)
         return self._celsius

    def get_fahrenheit(self):
        return ((9/5)*self._celsius)+32
        

#method 1:
t1=Temperature()
t1.set_celsius(32)
print(t1.get_celsius())

#method-2
t2=Temperature()
t2.set_fahrenheit(1)
(t2.get_fahrenheit())
```

```python
"""
Create a class Laptop with private variable _battery_level.

Write set_battery() method which only accepts values 0to100.

Write get_battery() method that prints:

“Low Battery” if less than 20
“Battery OK” otherwise.
"""
class Laptop:
    def __init__(self):
        self._battery_level=None

    def set_battery(self,battery_level):
        if battery_level>=0 and battery_level<=100:
          self._battery_level=battery_level
        else:
            print("enter bw 0 to 100")
    
    def get_battery(self):
        if self._battery_level==None:
            print("battery level is not proper")
        elif self._battery_level<20:
            print("Low Battery")
        else:
            print("Battery Ok")

l1=Laptop()
l1.set_battery(1088)
l1.get_battery()
    
```

```python
"""
 Create a class Movie with _rating.

Write set_rating() method that allows values only between 0 and 10.

Write get_rating() method:

If rating < 5 → print “Flop”
Else → print “Hit”.
"""

class Movie:
    def __init__(self):
        self._rating=None
    
    def set_rating(self,rating):
        if rating>=0 and rating<=100:
            self._rating=rating
        else:
            print("enter valid rating")

    def get_rating(self):
        if self._rating==None:
            print("rating not set")
        elif self._rating<5:
            print("Flop")
        else:
            print("Hit")

m1=Movie()
m1.set_rating(880)
m1.get_rating()
```
