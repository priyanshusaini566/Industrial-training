# question-2

```python
"""
Create a class Employee with private variable _age.

Write a setter method that accepts only age greater than 18.
If user tries to set less than 18, print “Invalid age”.
Write a getter method to print the age.
"""
class Employee:
    def __init__(self):
        self._age=None
    def setter(self,age):
        if age<18:
            print("invalid age")
        else:
            self._age=age

    def getter(self):
        print(self._age)

e1=Employee()
age=int(input("enter the age : "))
e1.setter(age)
e1.getter()
```

```python
"""
Create a class BankAccount with private variable _balance.

Start with balance = 0.
Write a method set_balance() to deposit money (only positive).
Write get_balance() to display the current balance.
Try depositing and printing the final balance.
"""
class BankAccount:
    def __init__(self):
      self._balance=0
    def set_balance(self,amount):
        if amount<0:
            print("enter positive amount!!")
        else:
           self._balance+=amount
    def get_balance(self):
        print(self._balance)

b1=BankAccount()
amt=int(input("Enter the amount to deposit : "))
b1.set_balance(amt)
b1.get_balance()
```

# question-4

```python
"""
Create a class Product with _price and _discount.

Use set_price() and set_discount() to assign values.
Write a get_final_price() method to return final price after discount. Formula: price - (price * discount / 100)
Display the final price.
"""

class Product:
    def set_price(self,amt):
        self._amt=amt
    def set_discount(self,discount):
        self._discount=discount
    def get_final_price(self):
        return self._amt-((self._amt*self._discount)/100)
    
p1=Product()
price=int(input("enter the price : "))
discount=int(input("enter the discount in % : "))
p1.set_price(price)
p1.set_discount(discount)
print("the final price will be : ",p1.get_final_price())
    
```
