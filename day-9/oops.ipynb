```python
# instance variable
class person:
  def getname(self):
    self.name="Priyanshu saini"
  def setname(self):
    return self.name
```

```python
p1=person()
```

```python
p1.getname()
```

```python
p1.setname()
```

```python
#getter and setter
class again:
    def setter(self,name):
        self.name=name
    def getter(self):
        return self.name
```

```python
a=again()
```
