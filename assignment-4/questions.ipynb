```python
# Write a Python function to find the key with the maximum value in a dictionary.
def max(d):
    max=0
    new=0
    for item in d:
        if d[item]>max:
            max=d[item]
            new=item

    print("the key with max value is : ",new)

d={1:200,3:49,5:6,7:8000}
max(d)



```

```python
# Write a Python program to reverse keys and values in a dictionary.
d={1:2,3:4,5:6,7:8}
l1=[]
l2=[]
for item in d:
    l1.append(item)
for item in d:
    l2.append(d[item])

l1.reverse()
l2.reverse()

d1={}

for i in range(len(l1)):
    d1[l1[i]]=l2[i]

print(d1)
```

# question-15

```python
# Write a Python function to merge multiple dictionaries into one.
def merge(d1,d2,d3,d4):

  for key,value in d2.items():
    d1[key]=value
  for key,value in d3.items():
    d1[key]=value
  for key,value in d4.items():
    d1[key]=value
  return d1

d1={1:2,3:4}
d2={5:6,7:8}
d3={11:22,33:44,55:66}
d4={21:22,23:24,25:26}

merge(d1,d2,d3,d4)
```

# question-19

```python
# Write a Python program to flatten a nested dictionary into a single-level dictionary
d={'a':{'b':1,'c':2},'d':3}
flat={}

for key,value in d.items():
    if isinstance(value,dict):
        for k,v in value.items():
            flat[key+"_"+k]=v
    else:
        flat[key]=value

print("the flatten dictionary is :",flat)
```

# question-20

```python
# Write a Python program to find common keys and different keys between two dictionaries.
d1={'a':'b','c':'d',1:2}
d2={'c':'m',1:7,3:10}

l1=[]
l2=[]
common=[]

d_keys=[]
for item in d1.keys():
    l1.append(item)

for item in d2.keys():
    l2.append(item)

for i in range(len(l1)):
    if l1[i] in l2:
        common.append(l1[i])
    else:
        d_keys.append(l1[i])

for i in range(len(l2)):
     if l2[i] not in common:
        d_keys.append(l2[i])


print("the common elements are : ",common)
print("the different elements are : ",d_keys)





```
