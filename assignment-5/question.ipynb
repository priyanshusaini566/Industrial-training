# question-6

```python
#question-6
def prime(x):
  flag=True
  if x==2 or x==1:
    flag=True
  else:
   for i in range(2,x):
    if x%i==0:
      flag=False
      break
  if flag:
    print("Prime")
  else:
    print("Not Prime")

prime(10)
```

# question-7

```python
# question-7
def reverse(s):
  s1=""
  for i in range(len(s)):
    s1=s[i]+s1
  return s1

s=input("Enter a string : ")
reverse(s)
```

# question-8

```python
#question-8
def count(s):
  vo="aeiou"
  count=0
  for i in range(len(s)):
    if s[i] in vo:
      count+=1
  return count
s=input("enter the string : ")
count(s)
```

```python
#question-10
def unique(l):
  l1=[]
  for i in range(len(l)):
    if (l.count(l[i]))==1:
      l1.append(l[i])
    else:
      continue
  return l1
l=[1,2,3,3,3,4]
unique(l)
```
