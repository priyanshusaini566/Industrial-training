# question-26

```python
# Write a Python function that uses reduce() and a lambda function to find the GCD of a list of numbers.
from math import gcd
from functools import reduce
def find(l):
    return reduce(lambda a,b:gcd(a,b),l)

l=[24,36]
find(l)
```

# question-28

```python
# Write a Python function that takes a string and returns a new string with words sorted by their length using sorted() and lambda.
def sort(s):
    words=s.split()
    s2=sorted(words,key=lambda x:len(x))
    return " ".join(s2)

s='i am priyanshu saini'
sort(s)
```

# question-29

```python
# Write a Python function to generate a list of Fibonacci numbers using recursion and map().
def fibo(n):
    if n<=1:
        return n
    else:
        return fibo(n-1)+fibo(n-2)

def fibona(n):
    l=range(n)
    
    return list(map(lambda x:fibo(x),l))

n=int(input("enter the no. of elements : "))
print(fibona(n))
```
