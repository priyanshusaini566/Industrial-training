```python
"""
Create an abstract class Payment with method:

pay() → display different payment methods like UPI or Card
Conditions:

Payment amount should be greater than 0
Display payment success message after transaction
"""

from abc import ABC,abstractmethod

class Payment(ABC):
    def check(self,payment):
        if payment<=0:
            print("payment amount should be greater than 0")
            return False
        return True
    
    @abstractmethod
    def method(self):
        pass

class upi(Payment):
    def method(self,payment):
        if self.check(payment):
            print("upi payment happened successfully")

class card(Payment):
    def method(self,payment):
        if self.check(payment):
            print("card payment successfully")
u1=upi()
u1.method(-20)
c1=card()
c1.method(20)
```

```python
"""
Create a hybrid inheritance example using:

Person → store basic details
Employee → store employee ID and salary
Intern → store internship duration and project name
Conditions:

Salary should be positive
Internship duration should be at least 1 month
"""
class Person:
    def details(self, name, age):
        self._name = name
        self._age = age
        print("Name:", self._name)
        print("Age:", self._age)


class Employee(Person):
    def employee_details(self, emp_id, salary):
        self._emp_id = emp_id

        if salary > 0:
            self._salary = salary
            print("Employee ID:", self._emp_id)
            print("Salary:", self._salary)
        else:
            print("Salary should be positive")


class Intern(Person):
    def intern_details(self, duration, project):
        if duration >= 1:
            self._duration = duration
            self._project = project

            print("Duration:", self._duration, "month(s)")
            print("Project:", self._project)
        else:
            print("Internship duration should be at least 1 month")


class Trainee(Employee, Intern):
    pass


t1 = Trainee()

t1.details("Priyanshu", 20)
t1.employee_details("E101", 15000)
t1.intern_details(-3, "Weather App")
```

```python
"""
Create a Shape class with method:

area() → calculate area for square and rectangle using same method name
Conditions:

Values should be positive
Display invalid message for wrong input
"""
class Shape:
    def area(self, *args):
        if len(args) == 1:  
            side = args[0]

            if side > 0:
                print("Area of Square:", side * side)
            else:
                print("Invalid input")

        elif len(args) == 2:  
            length, breadth = args

            if length > 0 and breadth > 0:
                print("Area of Rectangle:", length * breadth)
            else:
                print("Invalid input")

        else:
            print("Invalid input")


s1 = Shape()

s1.area(5)   
s1.area(4, 6)   
s1.area(-2)   
s1.area(3, -1)
```

```python
"""
Create a Library Management System using OOP concepts.

Add books
Issue books
Return books
Display available books
Conditions:

Do not issue unavailable books
Book ID should be unique
"""
class Library:
    def __init__(self):
        self.books = {}

    def add_book(self, book_id, book_name):
        if book_id in self.books:
            print("Book ID should be unique")
        else:
            self.books[book_id] = {
                "name": book_name,
                "available": True
            }
            print(f"Book '{book_name}' added successfully")

    def issue_book(self, book_id):
        if book_id not in self.books:
            print("Book not found")
        elif not self.books[book_id]["available"]:
            print("Book is unavailable")
        else:
            self.books[book_id]["available"] = False
            print(f"Book '{self.books[book_id]['name']}' issued")

    def return_book(self, book_id):
        if book_id not in self.books:
            print("Book not found")
        else:
            self.books[book_id]["available"] = True
            print(f"Book '{self.books[book_id]['name']}' returned")

    def display_books(self):
        print("\nAvailable Books:")
        found = False

        for book_id, details in self.books.items():
            if details["available"]:
                print(f"ID: {book_id}, Name: {details['name']}")
                found = True

        if not found:
            print("No books available")


lib = Library()

lib.add_book(101, "Python Basics")
lib.add_book(102, "Data Structures")
lib.add_book(101, "OOP Concepts")  

lib.display_books()

lib.issue_book(101)
lib.issue_book(101)  

lib.display_books()

lib.return_book(101)

lib.display_books()
```

```python
"""
Create an ATM Machine project using encapsulation.

Deposit money
Withdraw money
Check account balance
Validate PIN
Conditions:

PIN should be 4 digits
Withdrawal should not exceed balance
Maximum 3 wrong PIN attempts allowed

"""
class ATM:
    def __init__(self, pin, balance=0):
        if len(str(pin)) != 4:
            raise ValueError("PIN should be 4 digits")

        self.__pin = str(pin)     
        self.__balance = balance   
        self.__attempts = 0

    def validate_pin(self, pin):
        if self.__attempts >= 3:
            print("Account locked. Maximum attempts exceeded.")
            return False

        if str(pin) == self.__pin:
            self.__attempts = 0
            return True

        self.__attempts += 1
        print(f"Wrong PIN! Attempts left: {3 - self.__attempts}")

        if self.__attempts >= 3:
            print("Account locked.")

        return False

    def deposit(self, pin, amount):
        if self.validate_pin(pin):
            if amount > 0:
                self.__balance += amount
                print("Deposit Successful")
                print("Balance:", self.__balance)

    def withdraw(self, pin, amount):
        if self.validate_pin(pin):
            if amount > self.__balance:
                print("Insufficient Balance")
            else:
                self.__balance -= amount
                print("Withdrawal Successful")
                print("Balance:", self.__balance)

    def check_balance(self, pin):
        if self.validate_pin(pin):
            print("Current Balance:", self.__balance)


atm = ATM(1234, 5000)

atm.check_balance(1234)

atm.deposit(1234, 2000)

atm.withdraw(1234, 3000)

atm.withdraw(1234, 10000)  

atm.check_balance(1111)
atm.check_balance(2222)
atm.check_balance(3333)    
atm.check_balance(1234)
```
