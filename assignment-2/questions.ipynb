# question-8

```python
# Write a program that takes a number and checks if it is the same when reversed (e.g., 121, 454).
N=int(input("Enter the number : "))
num=N
count=0
while(N>0):
  r=N%10
  count=(count*10)+r
  N=N//10
if count==num:
  print("it is a palindrome number")
else:
  print("it is not a palindrome number")
```

```python
# Write a program that takes a number as input and finds the largest digit in it.
# 123678908754 => 9
N=int(input("Enter the number : "))
max=0
while(N>0):
  r=N%10
  if r>max:
      max=r

  N=N//10

print("the maximum digit is : ",max)
```

# question-11

```python
# Write a program that checks if a given number is an Armstrong number (e.g., 153 = 1³ + 5³ + 3³).
N=int(input("Enter the number : "))
sum=0
num=N
while(N>0):
  r=N%10
  sum=sum+(r**3)
  N=N//10

if sum==num:
  print("it is an armstrong number")
else:
  print("not an armstrong number")
```

# question-14

```python
"""
Write a program that takes an integer ( N ) and repeatedly applies the following rules:
If ( N ) is even, divide it by 2.
If ( N ) is odd, multiply it by 3 and add 1.
Repeat until ( N ) becomes 1.
"""
N=int(input("Enter the number : "))

while(N!=1):
  if N%2==0:
    N=N//2
    print(N)
  else:
    N=(N*3)+1
    print(N)
```
