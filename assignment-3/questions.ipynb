```python
# Check if a given string is a palindrome (ignoring case and spaces)
str=input("Enter the string : ")
s=""
length=len(str)
for i in range((length-1),-1,-1):
    s=s+str[i]
if s==str:
    print("the string",str, "is palindrome")
else:
    print(str," is not palindrome")
    
```

# question-11

```python
#Find the longest word in a sentence.
string=input("Enter the sentence : ")
s=string.split(sep=" ")
size=len(string.split(sep=" "))
max=0
for i in range(size):
  if len(s[i])>max:
    max=len(s[i])
for i in range(size):
  if len(s[i])==max:
    print("String ",s[i],"is largest")
```

```python
# Sort a list in ascending order without using sort().
l=[4,2,8,6,1,80,34,1]
for i in range(len(l)):
  for j in range(i+1,len(l)):
    if l[j]<l[i]:
      l[i],l[j]=l[j],l[i]

print(l)
```
