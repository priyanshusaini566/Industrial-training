# question-2

```python
# palindrome
str=input("Enter the string : ")
s=""
length=len(str)
for i in range((length-1),-1,-1):
    s=s+str[i]
if s==str:
    print("the string is palindrome")
else:
    print("it is not palindrome")
    
```
